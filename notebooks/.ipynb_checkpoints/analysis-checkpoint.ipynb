# Chicago Homicides Analysis (1965–1995)
**Dataset:** ICPSR 6399 — *Homicides in Chicago, 1965–1995*  
Block, C.R., Block, R.L., & Illinois Criminal Justice Information Authority (2005).  
https://doi.org/10.3886/ICPSR06399.v5

---
### Sections
1. Gun crimes before vs. after July 1968 — by gun type
2. Gun crimes committed by young male offenders (ages 15–40)
3. Offender trait co-occurrence and KMeans clustering
4. Interactive neighborhood heatmap by Chicago community area

```python
import os
import json
import warnings
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import folium
import pyreadstat
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({"figure.dpi": 120})
print("Imports OK")
```

```python
# ── Paths ──────────────────────────────────────────────────────────────
DATA_DIR     = os.path.join("..", "data")
PART1_PATH   = os.path.join(DATA_DIR, "06399-0001-Data.por")
PART2_PATH   = os.path.join(DATA_DIR, "06399-0002-Data.por")
GEOJSON_PATH = os.path.join(DATA_DIR, "chicago_community_areas.geojson")

CHICAGO_GEOJSON_URL = (
    "https://data.cityofchicago.org/api/geospatial/igwz-8jzy"
    "?method=export&type=GeoJSON"
)

for p in [PART1_PATH, PART2_PATH]:
    status = "OK" if os.path.exists(p) else "NOT FOUND"
    print(f"{status}  {p}")
```

---
## Data Loading

```python
df1, meta1 = pyreadstat.read_por(PART1_PATH)

# ── Build 4-digit year and add gun-type binary flags ───────────────────
df1["YEAR4"]  = pd.to_numeric(df1["INJYEAR"],  errors="coerce").astype("Int64") + 1900
df1["MONTH4"] = pd.to_numeric(df1["INJMONTH"], errors="coerce").astype("Int64")

# Gun-type flags: any non-zero value in the W column = that weapon type present
GUN_COLS = {
    "WHANDGUN": "Handgun",
    "WSHOTGUN": "Shotgun",
    "WRIFLE"  : "Rifle",
    "WAUTOMAT": "Automatic",
    "WGUNUNK" : "Gun (unknown type)",
}
for col in GUN_COLS:
    df1[col + "_FLAG"] = df1[col].gt(0)

# A record is a gun homicide if ANY gun type flag is set
df1["IS_GUN"] = df1[[c + "_FLAG" for c in GUN_COLS]].any(axis=1)

print(f"Part 1 (victim-level): {len(df1):,} records")
print(f"Gun homicides:         {df1['IS_GUN'].sum():,}")
print(f"Year range:            {df1['YEAR4'].min()} – {df1['YEAR4'].max()}")
```

```python
df2, meta2 = pyreadstat.read_por(PART2_PATH)

df2["YEAR4"]  = pd.to_numeric(df2["INJYEAR"],  errors="coerce").astype("Int64") + 1900
df2["MONTH4"] = pd.to_numeric(df2["INJMONTH"], errors="coerce").astype("Int64")

for col in GUN_COLS:
    df2[col + "_FLAG"] = df2[col].gt(0)

df2["IS_GUN"] = df2[[c + "_FLAG" for c in GUN_COLS]].any(axis=1)

print(f"Part 2 (offender-level): {len(df2):,} records")
print(f"Gun homicides:           {df2['IS_GUN'].sum():,}")
```

---
## Section 1 — Gun Crimes Before vs. After July 1968
Showing all gun types: handgun, shotgun, rifle, automatic, and unknown-type guns.  
The Gun Control Act was enacted October 1968 (with enforcement phases starting mid-1968).  
Charts: stacked bar by gun type (pre/post) + annual trend line with July 1968 marker.

```python
gun1 = df1[df1["IS_GUN"]].copy()

def classify_era(row):
    y, m = row["YEAR4"], row["MONTH4"]
    if pd.isna(y) or pd.isna(m):
        return None
    if y < 1968 or (y == 1968 and m <= 7):
        return "Pre–July 1968"
    return "Post–July 1968"

gun1["ERA"] = gun1.apply(classify_era, axis=1)

print(f"Total gun homicides: {len(gun1):,}")
print(gun1["ERA"].value_counts().to_string())
```

```python
ERA_ORDER  = ["Pre–July 1968", "Post–July 1968"]
era_subset = gun1[gun1["ERA"].notna()]

# Build stacked bar: count of each gun type per era
gun_type_era = pd.DataFrame({
    label: era_subset.groupby("ERA")[col + "_FLAG"].sum()
    for col, label in GUN_COLS.items()
}).reindex(ERA_ORDER)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# ── Left: stacked bar by gun type ──────────────────────────────────────
gun_type_era.plot(kind="bar", stacked=True, ax=axes[0],
                  colormap="tab10", edgecolor="white", width=0.5)
for container in axes[0].containers:
    axes[0].bar_label(container, label_type="center", fontsize=8,
                      fmt="%d", color="white", fontweight="bold")
axes[0].set_title("Gun Homicides by Firearm Type\nBefore vs. After July 1968", fontsize=13)
axes[0].set_xlabel("")
axes[0].set_ylabel("Number of Homicides")
axes[0].tick_params(axis="x", rotation=0)
axes[0].legend(title="Firearm Type", bbox_to_anchor=(1.01, 1), loc="upper left", fontsize=9)

# ── Right: annual trend line (all guns) ────────────────────────────────
annual = gun1[gun1["YEAR4"].notna()].groupby("YEAR4").size().reset_index(name="count")
axes[1].plot(annual["YEAR4"], annual["count"],
             marker="o", linewidth=2, color="#4878CF", markersize=4, zorder=3)
axes[1].fill_between(annual["YEAR4"], annual["count"], alpha=0.12, color="#4878CF")
axes[1].axvline(x=1968.58, color="red", linestyle="--", linewidth=1.8,
                label="July 1968", zorder=4)
axes[1].set_title("Annual Gun Homicides in Chicago\n1965–1995", fontsize=13)
axes[1].set_xlabel("Year")
axes[1].set_ylabel("Number of Homicides")
axes[1].legend(fontsize=10)
axes[1].xaxis.set_major_locator(mticker.MultipleLocator(5))

plt.suptitle("Section 1: Gun Crime Trends — All Firearm Types",
             fontsize=14, y=1.01, fontweight="bold")
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section1_gun_trends.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: section1_gun_trends.png")
```

```python
# Annual trend broken out by gun type
fig, ax = plt.subplots(figsize=(14, 6))

type_colors = plt.cm.tab10.colors
for i, (col, label) in enumerate(GUN_COLS.items()):
    subset = gun1[gun1[col + "_FLAG"] & gun1["YEAR4"].notna()]
    if len(subset) == 0:
        continue
    yr_counts = subset.groupby("YEAR4").size().reset_index(name="count")
    ax.plot(yr_counts["YEAR4"], yr_counts["count"],
            label=label, color=type_colors[i], linewidth=1.8, alpha=0.85)

ax.axvline(x=1968.58, color="red", linestyle="--", linewidth=1.8, label="July 1968", zorder=5)
ax.set_title("Annual Gun Homicides by Firearm Type — Chicago 1965–1995", fontsize=13)
ax.set_xlabel("Year")
ax.set_ylabel("Number of Homicides")
ax.xaxis.set_major_locator(mticker.MultipleLocator(5))
ax.legend(bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section1_by_type.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: section1_by_type.png")
```

## Section 2 — Gun Crimes by Young Male Offenders (Ages 15–40)
From Part 1, filtering on primary offender (`OFN1SEX == 1` = male, `OFN1AGE` in 15–40 range).  

**Age category coding (OAGE / OFN1AGE):**
| Code | Age Range |
|------|-----------|
| 3    | 15–19     |
| 4    | 20–24     |
| 5    | 25–29     |
| 6    | 30–34     |
| 7    | 35–39     |
| 8    | 40–44     |

*Note: age is stored as categories in the public-use file — exact birth year is not available.*

```python
# Age codes that cover approximately 15–40
# Code 3=15-19, 4=20-24, 5=25-29, 6=30-34, 7=35-39, 8=40-44
AGE_15_40 = {3.0, 4.0, 5.0, 6.0, 7.0, 8.0}
AGE_LABELS = {
    3.0: "15–19",
    4.0: "20–24",
    5.0: "25–29",
    6.0: "30–34",
    7.0: "35–39",
    8.0: "40–44",
}

gun_ym = gun1[
    gun1["OFN1SEX"].eq(1.0) &
    gun1["OFN1AGE"].isin(AGE_15_40)
].copy()

gun_ym["AGE_LABEL"] = gun_ym["OFN1AGE"].map(AGE_LABELS)

print(f"Gun homicides by young males (15–44): {len(gun_ym):,}")
print(f"As % of all gun homicides:            {len(gun_ym)/len(gun1)*100:.1f}%")
print("\nAge group counts:")
print(gun_ym["AGE_LABEL"].value_counts().reindex(AGE_LABELS.values()).to_string())
```

```python
ordered_ages = list(AGE_LABELS.values())

pivot = (
    gun_ym[gun_ym["YEAR4"].notna()]
    .groupby(["YEAR4", "AGE_LABEL"])
    .size()
    .unstack(fill_value=0)
    .reindex(columns=ordered_ages, fill_value=0)
)

fig, ax = plt.subplots(figsize=(16, 6))
pivot.plot(kind="bar", stacked=True, ax=ax,
           colormap="tab10", edgecolor="none", width=0.85)

if 1968 in pivot.index:
    idx_1968 = list(pivot.index).index(1968)
    ax.axvline(x=idx_1968 + 0.5, color="red", linestyle="--",
               linewidth=1.8, label="July 1968", zorder=5)

ax.set_title(
    "Gun Homicides by Young Male Offenders (Ages 15–44)\nBy Age Group and Year, Chicago 1965–1995",
    fontsize=13
)
ax.set_xlabel("Year")
ax.set_ylabel("Number of Homicides")
xtick_labels = [
    str(int(y)) if (int(y) % 5 == 0 or int(y) == 1968) else ""
    for y in pivot.index
]
ax.set_xticklabels(xtick_labels, rotation=45, ha="right")
ax.legend(title="Age Group", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section2_young_males.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: section2_young_males.png")
```

```python
age_share = gun_ym["AGE_LABEL"].value_counts().reindex(ordered_ages)

fig, ax = plt.subplots(figsize=(7, 7))
wedges, texts, autotexts = ax.pie(
    age_share.values,
    labels=age_share.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=plt.cm.tab10.colors[:len(age_share)],
    wedgeprops={"edgecolor": "white", "linewidth": 1.5}
)
for t in autotexts:
    t.set_fontsize(10)
ax.set_title(
    "Age Group Share Among Young Male Gun Offenders\nChicago 1965–1995",
    fontsize=12
)
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section2_age_share.png"), dpi=150, bbox_inches="tight")
plt.show()
```

## Section 3 — Offender Trait Analysis (Young Male Gun Offenders)
Uses **Part 2 (offender-level)** — one record per offender, no duplicated rows.

**Traits examined:**
- `GANG` — incident was gang-related (0=no, 1=yes)
- `DOMESTIC` — domestic relationship involved (0=no, >0=yes)
- `DRUG` — drug involvement (0=none, 1=drug sale, 2=drug use)
- `LIQUOR` — alcohol involvement (1=yes, 2=no)
- `PRIOROF` — prior offense on record (1=violent, 2=nonviolent)

Charts: frequency bars → co-occurrence heatmap → KMeans cluster profiles + PCA scatter.

```python
# Filter Part 2 for young male gun offenders
p2_ym = df2[
    df2["IS_GUN"] &
    df2["OSEX"].eq(1.0) &
    df2["OAGE"].isin(AGE_15_40)
].copy()

print(f"Young male gun offenders (Part 2): {len(p2_ym):,}")

# Build binary trait columns
p2_ym["TRAIT_GANG"]        = p2_ym["GANG"].eq(1.0).astype(int)
p2_ym["TRAIT_DOMESTIC"]    = p2_ym["DOMESTIC"].gt(0).astype(int)
p2_ym["TRAIT_DRUG_SALE"]   = p2_ym["DRUG"].eq(1.0).astype(int)
p2_ym["TRAIT_DRUG_USE"]    = p2_ym["DRUG"].eq(2.0).astype(int)
p2_ym["TRAIT_ALCOHOL"]     = p2_ym["LIQUOR"].eq(1.0).astype(int)
p2_ym["TRAIT_PRIOR_VIOL"]  = p2_ym["PRIOROF"].eq(1.0).astype(int)
p2_ym["TRAIT_PRIOR_NONV"]  = p2_ym["PRIOROF"].eq(2.0).astype(int)

TRAIT_COLS = [
    "TRAIT_GANG", "TRAIT_DOMESTIC", "TRAIT_DRUG_SALE",
    "TRAIT_DRUG_USE", "TRAIT_ALCOHOL", "TRAIT_PRIOR_VIOL", "TRAIT_PRIOR_NONV"
]
TRAIT_LABELS_MAP = {
    "TRAIT_GANG"       : "Gang-Related",
    "TRAIT_DOMESTIC"   : "Domestic Relationship",
    "TRAIT_DRUG_SALE"  : "Drug Sale Involved",
    "TRAIT_DRUG_USE"   : "Drug Use Involved",
    "TRAIT_ALCOHOL"    : "Alcohol Involved",
    "TRAIT_PRIOR_VIOL" : "Prior Violent Offense",
    "TRAIT_PRIOR_NONV" : "Prior Non-Violent Offense",
}

trait_df = p2_ym[TRAIT_COLS].copy().fillna(0)
trait_df.columns = [TRAIT_LABELS_MAP[c] for c in TRAIT_COLS]

print("\nTrait prevalence (% of young male gun offenders):")
print((trait_df.mean() * 100).round(1).sort_values(ascending=False).to_string())
```

```python
freq = (trait_df.mean() * 100).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(11, 5))
colors_bar = ["#D65F5F" if v >= 25 else "#4878CF" for v in freq.values]
bars = ax.barh(freq.index, freq.values, color=colors_bar, edgecolor="white", height=0.6)
for bar, val in zip(bars, freq.values):
    ax.text(val + 0.3, bar.get_y() + bar.get_height() / 2,
            f"{val:.1f}%", va="center", fontsize=10)
ax.set_xlabel("% of Young Male Gun Offenders")
ax.set_title(
    "Prevalence of Co-occurring Traits\nYoung Male Gun Offenders (Ages 15–44, Chicago 1965–1995)",
    fontsize=13
)
ax.set_xlim(0, min(100, freq.max() * 1.25))
ax.axvline(x=50, color="grey", linestyle=":", linewidth=1, label="50% line")
ax.legend(fontsize=9)
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section3_trait_frequency.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: section3_trait_frequency.png")
```

```python
corr = trait_df.corr()
mask_upper = np.triu(np.ones_like(corr, dtype=bool))

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(
    corr,
    mask=mask_upper,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1, vmax=1,
    linewidths=0.5,
    ax=ax,
    annot_kws={"size": 9}
)
ax.set_title(
    "Trait Co-occurrence Correlation Matrix\nYoung Male Gun Offenders (Ages 15–44, Chicago 1965–1995)",
    fontsize=13, pad=12
)
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section3_cooccurrence.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: section3_cooccurrence.png")
```

```python
N_CLUSTERS = 4  # adjust as desired (try 3–6)

X        = trait_df.values
scaler   = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init=10)
labels = kmeans.fit_predict(X_scaled)
trait_df["Cluster"] = labels

cluster_sizes = trait_df["Cluster"].value_counts().sort_index()
profile_cols  = list(TRAIT_LABELS_MAP.values())
profile = (
    trait_df.groupby("Cluster")[profile_cols].mean() * 100
)
profile.index = [f"Cluster {i}  (n={cluster_sizes[i]:,})" for i in profile.index]

fig, ax = plt.subplots(figsize=(14, 5))
profile.T.plot(kind="bar", ax=ax, colormap="Set2", edgecolor="white", width=0.75)
ax.set_title(
    f"Offender Trait Profiles by KMeans Cluster (k={N_CLUSTERS})\n"
    "Young Male Gun Offenders — Chicago 1965–1995",
    fontsize=13
)
ax.set_xlabel("Trait")
ax.set_ylabel("% of Offenders in Cluster")
ax.tick_params(axis="x", rotation=30)
ax.legend(title="Cluster", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section3_kmeans_clusters.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: section3_kmeans_clusters.png")
```

```python
pca    = PCA(n_components=2, random_state=42)
coords = pca.fit_transform(X_scaled)
pca_df = pd.DataFrame(coords, columns=["PC1", "PC2"])
pca_df["Cluster"] = labels

fig, ax = plt.subplots(figsize=(8, 6))
for c in sorted(pca_df["Cluster"].unique()):
    sub = pca_df[pca_df["Cluster"] == c]
    ax.scatter(sub["PC1"], sub["PC2"],
               label=f"Cluster {c}", alpha=0.35, s=16, linewidths=0)
ax.set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% variance)")
ax.set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% variance)")
ax.set_title("PCA Projection of Offender Trait Clusters", fontsize=12)
ax.legend()
plt.tight_layout()
fig.savefig(os.path.join(DATA_DIR, "section3_pca.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: section3_pca.png")
```

## Section 4 — Chicago Neighborhood Heatmap
Interactive `folium` choropleth of gun homicide concentration by community area (1–77).  
GeoJSON fetched from the Chicago Data Portal and cached locally.  

Adjust `MAP_YEAR_START` and `MAP_YEAR_END` to focus on any sub-period.

```python
# ── Time range for the map ─────────────────────────────────────────────
MAP_YEAR_START = 1965
MAP_YEAR_END   = 1995

if not os.path.exists(GEOJSON_PATH):
    print("Fetching Chicago community areas GeoJSON...")
    resp = requests.get(CHICAGO_GEOJSON_URL, timeout=30)
    resp.raise_for_status()
    with open(GEOJSON_PATH, "w") as f:
        json.dump(resp.json(), f)
    print(f"Cached to {GEOJSON_PATH}")
else:
    print(f"Using cached GeoJSON: {GEOJSON_PATH}")

with open(GEOJSON_PATH) as f:
    chicago_geo = json.load(f)

sample_props = chicago_geo["features"][0]["properties"]
print("GeoJSON feature properties:", sample_props)
```

```python
# Auto-detect the community area number and name keys in the GeoJSON
GEO_KEY = next(
    (k for k in sample_props if "area" in k.lower() and "num" in k.lower()),
    next((k for k in sample_props if "area" in k.lower()), list(sample_props)[0])
)
NAME_KEY = next((k for k in sample_props if "community" in k.lower()), None)
print(f"Area number key: '{GEO_KEY}'   Name key: '{NAME_KEY}'")

# Aggregate gun homicides by community area
map_gun = gun1[
    gun1["YEAR4"].notna() &
    gun1["YEAR4"].between(MAP_YEAR_START, MAP_YEAR_END) &
    gun1["COMAREA"].notna()
].copy()

map_gun["COMAREA_STR"] = map_gun["COMAREA"].astype(float).astype(int).astype(str)
crime_by_area = map_gun.groupby("COMAREA_STR").size().reset_index(name="gun_homicides")

print(f"\nGun homicides {MAP_YEAR_START}–{MAP_YEAR_END}: {crime_by_area['gun_homicides'].sum():,}")
print("Top 10 community areas:")
print(crime_by_area.sort_values("gun_homicides", ascending=False).head(10).to_string(index=False))
```

```python
chicago_map = folium.Map(
    location=[41.85, -87.65],
    zoom_start=11,
    tiles="CartoDB positron"
)

folium.Choropleth(
    geo_data=chicago_geo,
    name="Gun Homicides by Community Area",
    data=crime_by_area,
    columns=["COMAREA_STR", "gun_homicides"],
    key_on=f"feature.properties.{GEO_KEY}",
    fill_color="YlOrRd",
    fill_opacity=0.75,
    line_opacity=0.4,
    legend_name=f"Gun Homicides ({MAP_YEAR_START}–{MAP_YEAR_END})",
    nan_fill_color="lightgrey",
    highlight=True,
).add_to(chicago_map)

# Add tooltips with community name and count
area_lookup = crime_by_area.set_index("COMAREA_STR")["gun_homicides"].to_dict()
for feature in chicago_geo["features"]:
    props   = feature["properties"]
    area_n  = str(props.get(GEO_KEY, ""))
    name    = props.get(NAME_KEY, f"Area {area_n}") if NAME_KEY else f"Area {area_n}"
    count   = area_lookup.get(area_n, 0)
    props["_tooltip"] = f"{name}: {count:,} gun homicides"

folium.GeoJson(
    chicago_geo,
    style_function=lambda f: {"fillOpacity": 0, "weight": 0},
    tooltip=folium.GeoJsonTooltip(
        fields=["_tooltip"], aliases=[""], labels=False
    )
).add_to(chicago_map)

folium.LayerControl().add_to(chicago_map)

map_path = os.path.join(DATA_DIR, "section4_chicago_heatmap.html")
chicago_map.save(map_path)
print(f"Interactive map saved: {map_path}")
chicago_map
```
